We will aim to free up space by removing already processed files in a safe way. 

1 - Files will be analysed by going through an excel spreadsheet (i.e., on a per-experiment basis)

2 - We will build routines to calculate out of focus frames from the jumpcorrected file, and save those as an csv sheet in the folder.

3 - We will also leave a file in the folder with the first-last from the excel file, which will detail the first-last frames to process.

4 - We will generate routines to remove jumpCorrected files in a safe way, i.e., if the jumpCorrected-mc file is present and the files mentioned in 2-3 are present. A dry run option will be enabled by default, which will also calculate the size of removed stuff. 

```python
import pandas as pd
import os


import numpy as np
import sys
sys.path.append('../src')
from file_utilities import generate_oof_labels, compute_total_space, safe_remove_jumpCorrected, safe_remove_motionCorrected
```

```python
#Variables
processedFolder = 'processedMovies'
jumpcorrected_file = '1-jumpcorrected.tif'
motioncorrected_file = '1-jumpcorrected-mc.tif'
RAW_FILE = 'Image_001_001.raw'
JSON_FILE = 'processing_parameters.json'

oof_labels_file = 'out_of_focus_labels.csv'
```

```python
EXCEL_SPREASHEET = '../Myo15_IHCs.xlsx'

data = pd.read_excel(EXCEL_SPREASHEET)

#Change drive if necessary
#Mac os
data['Folder'] = data['Folder'].str.replace('\\', '/')
drive  = '/Volumes/Data'
data['Folder'] = data['Folder'].str.replace('Z:', drive)
```

```python
folders = (data['Folder'].str[:] + '/' + processedFolder).tolist()

_ = compute_total_space(folders,file_name=jumpcorrected_file)

_ = compute_total_space(folders,file_name=motioncorrected_file)
```

```python
import tqdm
for index, row in tqdm.tqdm(data.iterrows(), total=len(data)):
    first_last = row['first-last'].split('-')
    first,last = int(first_last[0]), int(first_last[1])
    try:
        nChannels = int(row['nChannels'])
    except KeyError:
        nChannels = 1  # Default to 1 if not specified or invalid
    try:
        spatialGaussian = float(row['SpatialGaussian'])
    except KeyError:
        spatialGaussian = 2.0 # Default to 2.0 if not specified or invalid
    try:
        temporalGaussian = float(row['TemporalGaussian'])
    except KeyError:
        temporalGaussian = 2.0 # Default to 2.0 if not specified or invalid
        

    folder_path = row['Folder']

    generate_oof_labels(os.path.join(folder_path, processedFolder, jumpcorrected_file),os.path.join(folder_path, processedFolder, oof_labels_file),first,last,nChannels,spatialGaussian,temporalGaussian,save_parameters=True,dry_run=True)
```

```python
safe_remove_jumpCorrected(data['Folder'].tolist(),dry_run=True)  # Set dry_run=False to actually delete files
```

```python
safe_remove_motionCorrected(folders,dry_run=True)  # Set dry_run=False to actually delete files
```
